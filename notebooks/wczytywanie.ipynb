```python
from pathlib import Path
import pandas as pd

# Sprawdzenie relatywnych ścieżek zależnie od katalogu roboczego notebooka
mozliwe_sciezki = [
    Path("../data/wynajmy.csv"),
    Path("data/wynajmy.csv"),
    Path("wynajmy.csv"),
]

sciezka_pliku = next((s for s in mozliwe_sciezki if s.is_file()), None)

if sciezka_pliku is None:
    raise FileNotFoundError("Nie znaleziono pliku wynajmy.csv w katalogu data/ ani w bieżącym.")

df = pd.read_csv(sciezka_pliku)
print(f"Pomyślnie wczytano zbiór danych ze ścieżki: {sciezka_pliku}")
print(f"Rozmiar zbioru: {df.shape[0]} wierszy, {df.shape[1]} kolumn")
```
